```python
import requests
from bs4 import BeautifulSoup
def get_webpage_title(url):
  #'''fixed webpage ka title scrape karta hai aur file m save krta hai """"
  try:
    # Webpage ka HTML content fetch karna
    response = requests.get(url)
    response.raise_for_status()# Agr error howi tou expectation raise krega
    #HTML ko parse karna
    soup = BeautifulSoup(response.text,'html.parser')
    #Title tag nikalna
    title = soup.title.string if soup.title else "No title found"
    print(f"Webpage URL: {url}")
    print(f"Title: {title}")
    #result ko file m save krna
    with open("webpage_title.txt","w",encoding="utf-8") as file:
      file.write(f"Webpage URL: {url}\n")
      file.write(f"Title: {title}\n")
    print("\nTitle successfully save to 'webpage_title.txt'")
  except Exception as e:
    print(f"An error occured: {e}")

#fixed webpage url
url = "https://www.geeksforgeeks.org" # Changed URL to a working one
get_webpage_title(url)
```

TASK NO #3 &&&&&   OPTION 1 MOVED JPG FILES


```python
import os
import shutil

def move_jpg_files(source_folder, destination_folder):
    # Create the destination folder if it doesn't exist
    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)

    move_count = 0
    if not os.path.exists(source_folder):
        print(f"Source folder '{source_folder}' does not exist. No files to move.")
        return

    for filename in os.listdir(source_folder):
        if filename.endswith(".jpg"):
            source_path = os.path.join(source_folder, filename)
            destination_path = os.path.join(destination_folder, filename) # Corrected destination path
            shutil.move(source_path, destination_path) # Corrected typo from shuthil to shutil
            print(f"Moved: {filename}")
            move_count += 1 # Corrected variable name from moved_count to move_count
    print(f"\nTotal {move_count} .jpg files move hogae hai.") # Corrected f-string syntax

source_folder = "/content/source_folder"
destination_folder = "/content/destination_folder"

# Create source and destination folders if they don't exist
os.makedirs(source_folder, exist_ok=True)
os.makedirs(destination_folder, exist_ok=True)

# Create some dummy .jpg files in the source folder for demonstration
with open(os.path.join(source_folder, "image1.jpg"), "w") as f: f.write("dummy content")
with open(os.path.join(source_folder, "image2.jpg"), "w") as f: f.write("dummy content")
with open(os.path.join(source_folder, "document.txt"), "w") as f: f.write("text content")

move_jpg_files(source_folder, destination_folder)
```

EMAIL ADDRESS EXTRACTED KRNE KE LIYE


```python
import re
def extract_emails(input_file, output_file):
  """This function extracts emails from an input file and saves them to an output file."""
  try:
    with open(input_file, 'r') as file:
      text = file.read()
      emails_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[a-zA-Z]{2,}'
      emails = re.findall(emails_pattern, text)
      with open(output_file, 'w') as output:
        for email in emails:
          output.write(email + '\n')
      print(f"Emails successfully extracted and saved to {output_file}")
  except FileNotFoundError:
    print(f"Input file '{input_file}' not found.")
  except Exception as e:
    print(f"An error occured: {e}")

# Example usage (moved outside the function and except block)
sample_text ="""
contact us at infp@example.com or support@test.org
Rabisa ka email hai rabia123@gmail.com
second contact : admin@company.pk
"""

with open("sample.txt", "w") as file:
  file.write(sample_text)

input_file = "sample.txt"
output_file = "extracted_emails.txt"
extract_emails(input_file, output_file)
```
