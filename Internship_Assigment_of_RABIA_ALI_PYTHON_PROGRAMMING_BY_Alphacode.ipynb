```python
import random

#list of words
words = ["apple", "mango","guava","banana","orange","lychee","pineapple"]
#Random words choose
word = random.choice(words)

guessed_letters = []
wrong_guesses = 0
max_wrong_guesses = 6

print("Welcome to Hangman Game")

while wrong_guesses < max_wrong_guesses:
  display_word = ""
  for letter in word:
    if letter in guessed_letters:
      display_word += letter
    else:
      display_word += "_"
  print(display_word)

  if "_" not in display_word:
    print("Congratulations! You guessed the word:", word)
    break

  guess = input("Guess a letter: ").lower()

  if len(guess) != 1 or not guess.isalpha():
    print("Please enter a single letter.")
    continue

  if guess in guessed_letters:
    print("You already guessed that letter. Try again.")
  elif guess in word:
    print(f"Good guess! '{guess}' is in the word.")
    guessed_letters.append(guess)
  else:
    print(f"Sorry, '{guess}' is not in the word.")
    wrong_guesses += 1
    guessed_letters.append(guess)

  if wrong_guesses == max_wrong_guesses:
    print("Game Over! You ran out of guesses.")
    print("The word was:", word)
    break
```

```python
stock_price ={
    "AAPL" : 180,
    "TSLA" : 250,
    "AMZN" : 175,
    "GOOGL" : 140,
    "MSFT" : 420
}
portfolio = {} # user ka purchase kia howa stock store hoga yaha
total_investment = 0
print("Welcome to the Stock Portfolio Tracker!")
print("Available stock:", list(stock_price.keys()))
while True :
  stock = input("\nEnter stock name (or type'done' to finish):").upper()
  if stock == "Done":
    break
  if stock not in stock_price:
    print("Invalid stock name. Please try again.")
    continue
  try:
    quantity = int(input(f"Enter the quantity{stock}:"))
    if quantity <= 0:
       print("Invalid quantity. Please try again.")
       continue
    cost = stock_price[stock] * quantity
    if stock in portfolio:
      portfolio[stock] += quantity
      total_investment += cost
    else:
      portfolio[stock] = quantity
      total_investment += cost
    print(f"Added {quantity} shares of {stock} (${cost} total)")
  except ValueError: # Added except block to fix SyntaxError
    print("Invalid quantity. Please enter a numerical value.")
    continue

# Final Summary - Moved outside the while loop to execute once
print("\n=====Portfolio Summary=====")
for stock , qty in portfolio.items():
  price = stock_price[stock]
  print(f"{stock}: {qty} shares x ${price} = ${qty * price}")
print(f"\nTotal Investment: ${total_investment}")

# SAVE IN FILE - Moved outside the while loop to execute once
with open("portfolio.txt","w") as file:
  file.write("Stock portfolio Summary\n")
  file.write("========================\n")
  for stock, qty in portfolio.items():
    price = stock_price[stock]
    file.write(f"{stock}:{qty} shares X ${price} = ${qty * price}\n")
  file.write(f"\nTotal Investment: ${total_investment}\n")
print("Data saved to 'portfolio.txt'")
```
